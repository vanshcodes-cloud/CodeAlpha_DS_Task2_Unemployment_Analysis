## Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 📂 Loading the Dataset

Let's load the unemployment dataset into a Pandas DataFrame.

```python
df = pd.read_csv("Unemployment in India.csv")
```

### 📊 Statistical Summary

Let's generate descriptive statistics for the numerical columns.

```python
df.describe()
```

### ❓ Checking for Missing Values

Let's check whether the dataset contains any missing values.

```python
df.isnull().sum()
```

### 🧹 Handling Missing Values

Let's remove the remaining missing values from the dataset to ensure clean and reliable analysis.

```python
df = df.dropna().reset_index(drop=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
```

### 🔁 Checking for Duplicate Rows

Let's check whether the dataset contains any duplicate records.

```python
df.duplicated().sum()
```

### 🧹 Removing Duplicate Rows

Duplicate rows, if present, will be removed to keep the dataset clean.

```python
df = df.drop_duplicates()
```

### 📐 Updated Dataset Shape

Let's check the dataset dimensions after removing duplicate rows.

```python
print("Rows and Columns after removing duplicates:", df.shape)
```

## 🧹 Data Cleaning and Preparation

Before performing visual analysis, we will clean the column names and prepare the dataset for further analysis.

```python
df.columns = df.columns.str.strip()
```

### 🏷️ Renaming Columns

Let's rename the columns to shorter and easier-to-use names.

```python
df.rename(columns={
    "Estimated Unemployment Rate (%)": "Unemployment Rate",
    "Estimated Employed": "Employed",
    "Estimated Labour Participation Rate (%)": "Labour Participation Rate"
}, inplace=True)

df.columns
```

### 📅 Converting Date Column

Let's convert the Date column into a proper datetime format for time-based analysis.

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

### 🔍 Checking Data Types

Let's verify the updated data types after cleaning the dataset.

```python
df.dtypes
```

## 📊 Unemployment Rate Distribution

Let's visualize the distribution of unemployment rates in the dataset.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Unemployment Rate"],
    bins=20,
    kde=True
)

plt.title("Distribution of Unemployment Rate")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Frequency")

plt.show()
```

## 📈 Overall Unemployment Trend

Let's analyze how the unemployment rate changed over time across the dataset.

```python
monthly_unemployment = df.groupby("Date")["Unemployment Rate"].mean()

plt.figure(figsize=(10,5))

plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker="o"
)

plt.title("Overall Unemployment Rate Trend")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

## 🗺️ Region-wise Unemployment Analysis

Let's compare the average unemployment rate across different regions of India.

```python
region_unemployment = (
    df.groupby("Region")["Unemployment Rate"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=region_unemployment.values,
    y=region_unemployment.index
)

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.show()
```

## 🌾 Rural vs Urban Unemployment

Let's compare the average unemployment rate between rural and urban areas.

```python
area_unemployment = df.groupby("Area")["Unemployment Rate"].mean()

plt.figure(figsize=(7,5))

sns.barplot(
    x=area_unemployment.index,
    y=area_unemployment.values
)

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.xlabel("Area")
plt.ylabel("Average Unemployment Rate (%)")

plt.show()
```

## 🦠 COVID-19 Impact on Unemployment

Let's compare unemployment rates before and during the COVID-19 period to understand its impact on unemployment in India.

```python
pre_covid = df[df["Date"] < "2020-03-01"]["Unemployment Rate"].mean()

covid_period = df[df["Date"] >= "2020-03-01"]["Unemployment Rate"].mean()

print("Average Unemployment Rate Before COVID-19:", round(pre_covid, 2), "%")
print("Average Unemployment Rate During COVID-19:", round(covid_period, 2), "%")
```

## 📊 Before vs During COVID-19

The following visualization compares the average unemployment rate before and during the COVID-19 period.

```python
covid_comparison = pd.Series({
    "Before COVID-19": pre_covid,
    "During COVID-19": covid_period
})

plt.figure(figsize=(7,5))

sns.barplot(
    x=covid_comparison.index,
    y=covid_comparison.values
)

plt.title("Unemployment Rate Before vs During COVID-19")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")

plt.show()
```

## 📈 Employment Trend

Let's examine how the estimated number of employed people changed over time.

```python
monthly_employment = df.groupby("Date")["Employed"].mean()

plt.figure(figsize=(10,5))

plt.plot(
    monthly_employment.index,
    monthly_employment.values,
    marker="o"
)

plt.title("Average Estimated Employment Trend")
plt.xlabel("Date")
plt.ylabel("Estimated Employed")
plt.xticks(rotation=45)

plt.show()
```

## 👷 Labour Participation Rate Analysis

Let's analyze the average labour participation rate across different regions.

```python
labour_participation = (
    df.groupby("Region")["Labour Participation Rate"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=labour_participation.values,
    y=labour_participation.index
)

plt.title("Average Labour Participation Rate by Region")
plt.xlabel("Average Labour Participation Rate (%)")
plt.ylabel("Region")

plt.show()
```

## 🔗 Correlation Analysis

Let's examine the relationship between unemployment rate, employment, and labour participation rate.

```python
numeric_df = df[
    [
        "Unemployment Rate",
        "Employed",
        "Labour Participation Rate"
    ]
]

plt.figure(figsize=(8,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")
plt.show()
```

## 💡 Key Insights

- 📈 The unemployment rate varies across different regions of India.
- 🗺️ Some regions show considerably higher unemployment rates than others.
- 🌾 Unemployment rates differ between rural and urban areas.
- 🦠 The unemployment rate increased significantly during the COVID-19 period.
- 👷 Employment levels also changed during the COVID-19 period.
- 🔗 Unemployment, employment, and labour participation show relationships that can be observed through the correlation analysis.

## 📅 Extended COVID-19 Unemployment Analysis

The second unemployment dataset provides additional data for 2020. We will use it to analyze the unemployment trend during the extended COVID-19 period.

```python
extended_df = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("Extended Dataset Shape :", extended_df.shape)
extended_df.head()
```

### 🔍 Extended Dataset Overview

Let's inspect the structure and data types of the extended unemployment dataset.

```python
extended_df.info()
```

### 📈 Extended Unemployment Trend

Let's visualize the unemployment rate trend using the extended 2020 dataset.

```python
extended_df.columns = extended_df.columns.str.strip()

extended_df["Date"] = pd.to_datetime(
    extended_df["Date"],
    dayfirst=True
)

extended_unemployment = (
    extended_df.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
)

plt.figure(figsize=(10,5))

plt.plot(
    extended_unemployment.index,
    extended_unemployment.values,
    marker="o"
)

plt.title("Extended Unemployment Rate Trend in 2020")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

### 🗺️ Region-wise Extended Analysis

Let's compare the average unemployment rate across regions using the extended 2020 dataset.

```python
region_extended = (
    extended_df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=region_extended.values,
    y=region_extended.index
)

plt.title("Average Unemployment Rate by Region - Extended Dataset")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.show()
```
